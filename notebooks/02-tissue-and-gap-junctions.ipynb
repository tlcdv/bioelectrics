# 2. A tissue, and the boundary a domain draws for itself

Cells are coupled by gap junctions that close when the voltage across them is large.
A depolarized domain therefore isolates itself, which is what lets it persist beside
healthy tissue.

```python
import torch
import matplotlib.pyplot as plt
from bioelectrics.ionic import IonicTissue, ChannelExpression, K_IR_CAM, CAT_DEP_CAM
from bioelectrics.attractors import find_steady_state
from bioelectrics.viz import plot_ionic_vmem_map

tissue = IonicTissue(channels=[ChannelExpression(K_IR_CAM, 1e-16),
                               ChannelExpression(CAT_DEP_CAM, 1e-16)])

v0 = torch.full((1, 10, 10), -0.078, dtype=torch.float64)
v0[:, 4:7, 4:7] = -0.003
state = tissue.with_vmem(tissue.initial_state((10, 10)), v0)
```

```python
ss = find_steady_state(tissue, state)
print('stable:', ss.stable[0])

fig, ax = plt.subplots(figsize=(4.5, 4))
im, n_closed = plot_ionic_vmem_map(ax, ss.vmem[0], v_center_mV=-45.0)
ax.set_title('{} closed gap junctions'.format(n_closed))
fig.colorbar(im, ax=ax, label='V_mem (mV)')
plt.show()
```

The dark lines are junctions whose open fraction has fallen. Nobody drew that
boundary. It follows from the voltage difference across each junction.
